```python
import numpy as np

# 딕셔너리에서 값 추출
p00 = result2["parameters"]["p00"]
p01 = result2["parameters"]["p01"]
p10 = result2["parameters"]["p10"]
p11 = result2["parameters"]["p11"]

P = np.array([
    [p00, p01, p10, p11],
    [p01, p00, p11, p10],
    [p10, p11, p00, p01],
    [p11, p10, p01, p00],
])

ideal = np.array([1.0, 0.0, 0.0, 0.0])

# PEC 계수: 역행렬을 명시적으로 만들지 않고 계산
gamma = np.linalg.solve(P, ideal)

# PEC 샘플링 확률 및 결과 가중치
G = np.abs(gamma).sum()
pec_probabilities = np.abs(gamma) / G
pec_weights = G * np.sign(gamma)

def calculate_alpha(lambda_scale):
    if not np.isfinite(lambda_scale) or lambda_scale < 1:
        raise ValueError("lambda_scale은 1 이상의 유한한 값이어야 합니다.")

    alpha = (1 - lambda_scale) * gamma + lambda_scale * ideal

    if np.any(alpha < -1e-12):
        raise ValueError(f"음수 α가 있어 확률로 샘플링할 수 없습니다: {alpha}")

    # 부동소수점 반올림 오차만 정리
    alpha = np.maximum(alpha, 0)
    return alpha / alpha.sum()

print("gamma:", gamma)
print("PEC sampling probabilities:", pec_probabilities)
print("PEC weights:", pec_weights)
print("alpha(lambda=2):", calculate_alpha(2))
print("alpha(lambda=3):", calculate_alpha(3))
```

```python
import qpe_kriss

result = qpe_kriss.analyze_benchmark(
    "benchmark_e914570b92384ab69882e363e4f485c7"
)

import numpy as np
import matplotlib.pyplot as plt

analysis = result["analysis"]
panels = analysis["distributions"]

x = np.arange(8)
labels = [f"{i:03b}" for i in range(8)]

def standard_errors(method):
    covariance = analysis.get("covariances", {}).get(method)
    if covariance is None:
        return np.zeros(8)

    return np.sqrt(
        np.maximum(np.diag(np.asarray(covariance)), 0)
    )

def coordinates(values):
    values = np.asarray(values, dtype=float)
    angles = 2 * np.pi * np.arange(8) / 8

    return (
        float(values @ np.cos(angles)),
        float(values @ np.sin(angles)),
    )

# analyze_benchmark가 만드는 기존 그래프는 닫기
plt.close(result["figure"])
```

```python
# ==============================================================================
# Phase Vector (2D Coordinate) & Euclidean Distance to IDEAL (Delta)
# ==============================================================================
import numpy as np
import pandas as pd

# ── 1. 분석 대상 기법 정의 ────────────────────────────────────
eval_methods = ["IDEAL", "RAW", "Matrix Inv", "ZNE", "PEC"]

# ── 2. 좌표 및 유클리드 거리(Delta) 계산 ───────────────────────
# panels 딕셔너리(또는 analysis["distributions"])에서 IDEAL 기준 좌표 추출
ideal_vec = coordinates(panels["IDEAL"])

results = []
print("=" * 68)
print(f"{'Method':<14} | {'Phase Vector (⟨cos⟩, ⟨sin⟩)':<26} | {'Distance to IDEAL (Δ)':<20}")
print("=" * 68)

for method in eval_methods:
  # 각 기법의 분포로부터 2D 위상 기댓값 벡터 계산
  vec = coordinates(panels[method])

  # IDEAL과의 유클리드 거리 (Euclidean Distance: Delta)
  delta = np.linalg.norm(vec - ideal_vec)

  results.append({
      "Method": method,
      "⟨cos(2πφ̂)⟩": vec[0],
      "⟨sin(2πφ̂)⟩": vec[1],
      "Distance (Δ)": delta,
  })

  # 터미널 / 텍스트 포맷 출력
  print(f"{method:<14} | ({vec[0]:>8.5f}, {vec[1]:>8.5f})      | {delta:>12.5e}")

print("=" * 68)

# ── 3. 노트북용 요약 데이터프레임 (DataFrame) ─────────────────
df_results = pd.DataFrame(results).set_index("Method")
display(
    df_results.style.format({
        "⟨cos(2πφ̂)⟩": "{:.5f}",
        "⟨sin(2πφ̂)⟩": "{:.5f}",
        "Distance (Δ)": "{:.5e}",
    }).highlight_min(
        subset=["Distance (Δ)"], color="#d4edda"
    )  # 거리가 가장 짧은(개선된) 행 강조
)
```
